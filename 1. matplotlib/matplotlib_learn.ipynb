```python
# Movies Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.plot(oscar_years, oscar_revenue)
plt.title("Oscar Years vs Revenue(in $M)")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(oscar_years, oscar_revenue, label="Oscar Winners")
plt.plot(non_oscar_years, non_oscar_revenue, label="Non-Oscar Winners")
plt.title("Years vs Revenue(in $M)")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
plt.legend()
# plt.legend(loc="upper right")
# plt.legend(loc="upper left")
# plt.legend(loc="center")
```

```python
#Bar Charts 
movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.bar(years, oscar_revenue, color = "red")
plt.title("Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+10, str(oscar_revenue[i]), ha ="center")
```

```python
import numpy as np
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))   
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar Movies")

plt.title("Oscar Vs Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue")
plt.legend()
```

```python
#Scatter Plot 
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]

plt.scatter(age, bp, s = bp, cmap = 'OrRd', c = bp)
plt.title("Age Vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label = "BP")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], bp[i]), xytext = (age[i]+1, bp[i]))
plt.xlim(min(age), max(age) + 10)
plt.show(min(bp), max(bp) + 5)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humidity, label = "winter")
plt.scatter(summer_temp, summer_humidity, label = "summer")
plt.title("Summer Vs Winter Temp")
plt.xlabel("Temp in C")
plt.ylabel("Humidity in %")
plt.legend()
plt.show()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0, 0, 0.2, 0]

plt.pie(amounts, labels = expenses, autopct = "%1.1f%%",
       wedgeprops = {
           "edgecolor" : "black",
           "linewidth" : 1.5,
           "linestyle" : "--"
       },
        explode = explode,
        startangle = 0,
        shadow = True
       )
#plt.style.availabel
plt.style.use("ggplot")
plt.show()
```
