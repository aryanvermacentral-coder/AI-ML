```python
from PIL import Image
import matplotlib.pyplot as plt

# Read the image
img = Image.open('input_image.jpg')

# Print basic info
print("Image format:", img.format)
print("Image size (width x height):", img.size)
print("Image mode:", img.mode)

# Display the image
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Save the image (as required)
img.save('saved_image.jpg')
```

```python
from PIL import Image
import matplotlib.pyplot as plt

# Read the image
img = Image.open("input_image.jpg")

# DPI values as per assignment
dpi_values = [72, 150, 300, 600]

# Save the image at each DPI (metadata only)
for dpi in dpi_values:
    filename = f"output_dpi_{dpi}.jpg"
    img.save(filename, dpi=(dpi, dpi))

# Display in a 2x2 grid layout
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes = axes.flatten()

for ax, dpi in zip(axes, dpi_values):
    filename = f"output_dpi_{dpi}.jpg"
    reopened = Image.open(filename)
    ax.imshow(reopened)
    ax.set_title(f"DPI = {dpi}", fontsize=14, fontweight="bold")
    ax.axis("off")

plt.suptitle("Image Displayed at Different DPI Values", fontsize=16, fontweight="bold")
plt.subplots_adjust(hspace=0.3, wspace=0.1, top=0.90)
plt.show()
```
