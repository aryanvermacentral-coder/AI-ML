```python
#QUESTION 1 - Load and Analyze the recorded audio.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy import signal
```

```python
# Load audio
fs, x = wavfile.read("aryan_verma.wav")
```

```python
print("Sampling frequency:", fs, "Hz")
print("Number of samples:", len(x))
print("Duration:", len(x) / fs, "seconds")
```

```python
# Convert stereo to mono if necessary
if x.ndim > 1:
    x = np.mean(x, axis=1)
```

```python
# Convert to floating point
x = x.astype(float)
```

```python
# Normalize
x = x / np.max(np.abs(x))
```

```python
# Time axis
t = np.arange(len(x)) / fs
```

```python
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Recorded Speech Signal")
plt.grid()
plt.show()
```

```python
#QUESTION 2 - Global FFT
```

```python
N = len(x)
```

```python
# FFT
X = np.fft.rfft(x)
```

```python
# Frequency axis
freq = np.fft.rfftfreq(N, d=1/fs)
```

```python
# Magnitude spectrum
magnitude = np.abs(X)
```

```python
plt.figure(figsize=(12, 4))
plt.plot(freq, magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("Global FFT Spectrum")
plt.grid()
plt.show()
```

```python
threshold = 0.05 * np.max(magnitude)
```

```python
significant = np.where(magnitude > threshold)[0]
```

```python
lowest_frequency = freq[significant[0]]
highest_frequency = freq[significant[-1]]
```

```python
print("Lowest significant frequency:", lowest_frequency, "Hz")
print("Highest significant frequency:", highest_frequency, "Hz")
```

```python
#QUESTION 3 - STFT spectrogram
```

```python
#Window length 256
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Close all old figures
plt.close('all')

# Make sure x is valid
print("Sampling frequency:", fs)
print("Number of samples:", len(x))
print("Audio duration:", len(x) / fs, "seconds")

# STFT
f, tt, Zxx = signal.stft(
    x,
    fs=fs,
    window='hann',
    nperseg=256,
    noverlap=128
)

print("STFT shape:", Zxx.shape)
print("Frequency range:", f[0], "to", f[-1], "Hz")
print("Time range:", tt[0], "to", tt[-1], "seconds")

# Magnitude
magnitude = np.abs(Zxx)

# Create ONE figure and ONE axis
fig, ax = plt.subplots(figsize=(12, 5))

# Plot spectrogram
pcm = ax.pcolormesh(
    tt,
    f,
    magnitude,
    shading='auto'
)

ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Frequency (Hz)")
ax.set_title("STFT Spectrogram - Window Length 256")

# Colorbar belongs to THIS figure
fig.colorbar(pcm, ax=ax, label="Magnitude")

plt.tight_layout()
plt.show()
```

```python
#Window length - 1024
```

```python
# STFT - Window Length 1024

plt.close('all')

f, tt, Zxx = signal.stft(
    x,
    fs=fs,
    window='hann',
    nperseg=1024,
    noverlap=512
)

magnitude = np.abs(Zxx)

fig, ax = plt.subplots(figsize=(12, 5))

pcm = ax.pcolormesh(
    tt,
    f,
    magnitude,
    shading='auto'
)

ax.set_xlabel("Time (seconds)")
ax.set_ylabel("Frequency (Hz)")
ax.set_title("STFT Spectrogram - Window Length 1024")

fig.colorbar(
    pcm,
    ax=ax,
    label="Magnitude"
)

plt.tight_layout()
plt.show()
```
