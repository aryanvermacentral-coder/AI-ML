```python
#series (1D labelled array)

s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#indexing
print(s[0])
print(s[3])
```

```python
s = pd.Series([23,56,45,12,"abc"], index=["Adam", "Eve", "Bob", "Charlie", 25])
print(s)

#index
print(s["Adam"]) #23
# print(s[0])

#print  all index values
print(s.index)
```

```python
#vectorized operation on series

s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1+s2)

s1[0] = 100 #MUTABLE VLAUES
print(s1)

#IMMUTABLE SIZE
changed_s1 = s1.drop(0)
print(changed_s1)
```
